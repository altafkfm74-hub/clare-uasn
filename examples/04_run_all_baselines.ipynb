# Run CLARE and all ten baselines on a small world.

```
from clare.env.world import default_world
from clare.sim.engine import simulate
from clare.baselines import BASELINES
from clare.baselines.clare_protocol import CLAREProtocol

world = default_world(seed=0, n_nodes=20)
results = {'CLARE': simulate(CLAREProtocol(), world, duration_h=0.2, seed=0)}
for name, cls in BASELINES.items():
    results[name] = simulate(cls(), world, duration_h=0.2, seed=0)
for name, r in results.items():
    print(f'{name:15s}  PDR={r.pdr:.3f}  E/bit={r.energy_per_bit_nj:6.0f} nJ')
```
