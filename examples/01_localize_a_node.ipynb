# Localize a sensor via EKF

Uses `clare.core.localization`.

```
import numpy as np
from clare.core import localization as L

x_node = np.array([300., 400., 200.])
x_anc = np.array([0., 0., 0.])
t = L.toa_measurement(np.linalg.norm(x_node - x_anc), c0=1500.0, noise_std=1e-5)
print('ToA =', t, 's')
```
