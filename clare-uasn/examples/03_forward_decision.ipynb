# Choose a forwarder

Uses `clare.core.forwarder`.

```
from clare.core import forwarder as F

best = F.select_forwarder(
    candidates=[
        {'id': 'A', 'z': 100, 'E': 500, 'q': 5, 'E_Ntx': 1.2, 'eta': 0.9},
        {'id': 'B', 'z': 120, 'E': 800, 'q': 2, 'E_Ntx': 1.9, 'eta': 0.5},
    ],
    z_i=200,
)
print(best)
```
